```python
# Finding Outliers using IQR Method
outlier_counts = {}

# Define numeric columns, excluding ID which is not a meaningful numerical feature for outliers
numeric_columns = data.select_dtypes(include=[np.number]).columns.drop('ID', errors='ignore')

for column in numeric_columns:
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = data[
        (data[column] < lower_bound) |
        (data[column] > upper_bound)
    ]

    outlier_counts[column] = len(outliers)

outlier_counts
```

```python
outlier_df = pd.DataFrame(
    list(outlier_counts.items()),
    columns=["Feature", "Outlier_Count"]
)

outlier_df = outlier_df.sort_values(
    by="Outlier_Count",
    ascending=False
)

outlier_df
```

MODEL BUILDING

```python
features = [
    "Income",
    "Recency",
    "MntWines",
    "MntFruits",
    "MntMeatProducts",
    "MntFishProducts",
    "MntSweetProducts",
    "MntGoldProds",
    "NumWebPurchases",
    "NumCatalogPurchases",
    "NumStorePurchases",
    "NumWebVisitsMonth",
    "Total_Spending",
    "Total_Purchases",
    "Total_Campaign_Acceptance"
]

X = data[features].copy()
```

We select important customer demographic, spending, purchasing and campaign features. These features represent customer behaviour and will be used to create customer groups.

```python
#Handling Missing Values
X = X.fillna(X.median())
```

Missing values are replaced with the median value of each feature. This gives us complete data for clustering

```python
# Scaling the Data
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)
```

```python
# Find the Number of Clusters
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

inertia = []

for k in range(2, 11):
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    kmeans.fit(X_scaled)
    inertia.append(kmeans.inertia_)
```

```python
plt.figure(figsize=(8,5))

plt.plot(range(2, 11), inertia, marker="o")

plt.title("Elbow Method")
plt.xlabel("Number of Clusters")
plt.ylabel("Inertia")

plt.show()
```

The Elbow Method helps us choose the suitable number of customer groups. We look for the point where the decrease in inertia starts becoming smaller.

K-MEANS MODEL

```python
# K-Means Clustering
kmeans = KMeans(
    n_clusters=4,
    random_state=42,
    n_init=10
)

kmeans_labels = kmeans.fit_predict(X_scaled)
```

```python
data["KMeans_Cluster"] = kmeans_labels
```

K-Means divides customers into groups based on their similarity. Customers within the same cluster have similar characteristics and purchasing behaviour.

Hierarchical Clustering Model


```python
from sklearn.cluster import AgglomerativeClustering

hierarchical = AgglomerativeClustering(
    n_clusters=4
)

hierarchical_labels = hierarchical.fit_predict(X_scaled)

data["Hierarchical_Cluster"] = hierarchical_labels
```

Hierarchical clustering groups similar customers step by step. It creates clusters based on the distance between customers.

Gaussian Mixture Model


```python
from sklearn.mixture import GaussianMixture

gmm = GaussianMixture(
    n_components=4,
    random_state=42
)

gmm_labels = gmm.fit_predict(X_scaled)

data["GMM_Cluster"] = gmm_labels
```

GMM creates customer groups based on probability. It is useful when a customer may have characteristics similar to more than one group.

DBSCAN MODEL

```python
from sklearn.cluster import DBSCAN

dbscan = DBSCAN(
    eps=1.5,
    min_samples=5
)

dbscan_labels = dbscan.fit_predict(X_scaled)

data["DBSCAN_Cluster"] = dbscan_labels
```

DBSCAN creates groups based on the density of customers. It can also identify unusual customers as noise or outliers.

BIRCH MODEL

```python
from sklearn.cluster import Birch

birch = Birch(
    n_clusters=4
)

birch_labels = birch.fit_predict(X_scaled)

data["Birch_Cluster"] = birch_labels
```

BIRCH is a clustering algorithm designed to efficiently group large datasets. It summarizes similar data points and creates customer clusters.

EVALUATION

```python
from sklearn.metrics import (
    silhouette_score,
    davies_bouldin_score,
    calinski_harabasz_score
)
```

```python
models = {
    "K-Means": kmeans_labels,
    "Hierarchical": hierarchical_labels,
    "GMM": gmm_labels,
    "DBSCAN": dbscan_labels,
    "BIRCH": birch_labels
}

results = []

for name, labels in models.items():

    # Ignore noise points for DBSCAN
    mask = labels != -1

    if len(set(labels[mask])) > 1:
        silhouette = silhouette_score(
            X_scaled[mask],
            labels[mask]
        )

        davies = davies_bouldin_score(
            X_scaled[mask],
            labels[mask]
        )

        calinski = calinski_harabasz_score(
            X_scaled[mask],
            labels[mask]
        )

    else:
        silhouette = np.nan
        davies = np.nan
        calinski = np.nan

    results.append([
        name,
        silhouette,
        davies,
        calinski
    ])

results_df = pd.DataFrame(
    results,
    columns=[
        "Model",
        "Silhouette Score",
        "Davies-Bouldin Score",
        "Calinski-Harabasz Score"
    ]
)

results_df
```

Silhouette Score,
Higher = better.
It tells us how well-separated the customer groups are.

Davies-Bouldin Score,
Lower = better.
It tells us how similar or overlapping the clusters are.

Calinski-Harabasz Score,
Higher = better.
It measures how clearly separated the clusters are.

COMPARING ALL MODELS

```python
results_df.sort_values(
    by="Silhouette Score",
    ascending=False
)
```

We compare all five clustering models using three evaluation metrics. A good model should have a high Silhouette Score, a low Davies-Bouldin Score and a high Calinski-Harabasz Score.

```python
import matplotlib.pyplot as plt
import numpy as np

# Model names
models = ["K-Means", "Hierarchical", "BIRCH", "GMM"]

# Original results
silhouette = np.array([0.287411, 0.268379, 0.241734, 0.080003])
davies_bouldin = np.array([1.893764, 1.839451, 1.905494, 2.393245])
calinski_harabasz = np.array([787.422964, 703.825304, 683.735550, 555.020264])

# Normalize each metric between 0 and 1
silhouette_norm = silhouette / silhouette.max()

# For Davies-Bouldin, lower is better, so invert it
davies_norm = davies_bouldin.min() / davies_bouldin

# For Calinski-Harabasz, higher is better
calinski_norm = calinski_harabasz / calinski_harabasz.max()

# X-axis positions
x = np.arange(len(models))
width = 0.25

# Create graph
plt.figure(figsize=(12, 6))

plt.bar(
    x - width,
    silhouette_norm,
    width,
    label="Silhouette Score"
)

plt.bar(
    x,
    davies_norm,
    width,
    label="Davies-Bouldin Score"
)

plt.bar(
    x + width,
    calinski_norm,
    width,
    label="Calinski-Harabasz Score"
)

plt.xlabel("Clustering Models")
plt.ylabel("Normalized Score")
plt.title("Clustering Model Performance Comparison")

plt.xticks(x, models)
plt.ylim(0, 1.1)
plt.legend()

plt.tight_layout()
plt.show()
```

“This graph compares the performance of different clustering models using three evaluation metrics. The scores are normalized to make the metrics comparable. K-Means shows the strongest overall performance, while GMM performs lower compared with the other models.”

Highest Silhouette Score (0.2874) → best cluster separation among these models.

Highest Calinski-Harabasz Score (787.42) → strongest overall separation/compactness according to this metric.

Its Davies-Bouldin score is 1.8938, slightly higher than Hierarchical's 1.8395, so Hierarchical is a little better on that one metric.

But considering the overall results, K-Means is the strongest choice

DBSCAN settings did not produce a valid cluster structure for these evaluation calculations.

Final K-Means Model

```python
# Training the Final Model
from sklearn.cluster import KMeans

final_kmeans = KMeans(
    n_clusters=4,
    random_state=42,
    n_init=10
)

final_labels = final_kmeans.fit_predict(X_scaled)

data["Final_Cluster"] = final_labels
```

```python
#Check the Final Clusters
data["Final_Cluster"].value_counts().sort_index()
```

“After comparing different clustering models, we selected K-Means as our final model. We trained it with four clusters and assigned each customer to one of these clusters. These final cluster labels will be used for customer segmentation and deployment.”

```python
# Check Final Model Performance
final_silhouette = silhouette_score(X_scaled, final_labels)

final_davies = davies_bouldin_score(
    X_scaled,
    final_labels
)

final_calinski = calinski_harabasz_score(
    X_scaled,
    final_labels
)

print("Silhouette Score:", final_silhouette)
print("Davies-Bouldin Score:", final_davies)
print("Calinski-Harabasz Score:", final_calinski)
```

```python
# Save the Final Model
import joblib

joblib.dump(final_kmeans, "customer_segmentation_kmeans.pkl")
joblib.dump(scaler, "customer_segmentation_scaler.pkl")

print("K-Means model and scaler saved successfully!")
```
